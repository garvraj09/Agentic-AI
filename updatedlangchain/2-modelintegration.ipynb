```python
## Using Ollama (llama3.2) model :

from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")   ## we can use any other model of ollama also as per the requirements

response = model.invoke("Hello, how are you?")
print(response.content)
```

### Models Integration with OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
## Using OpenAI model :

from langchain.chat_models import init_chat_model
model=init_chat_model("gpt-4.1")
model
```

```python
## Using Groq model

from langchain.chat_models import init_chat_model
model = init_chat_model(
    "llama-3.3-70b-versatile",
    model_provider="groq"
)
model
```

```python
## invoke the model
response=model.invoke("Hello how are you")
response.content
```

```python
## Using Google Gemini Model :

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# Load the API key from .env
load_dotenv(override=True)

# Initialize the Gemini model
model = init_chat_model(
    "gemini-3.8-flash",
    model_provider="google_genai"
)

# Invoke the model
response = model.invoke("Hello, how are you?")

print(response.content)
```

### Google Gemini Model Integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response
```
