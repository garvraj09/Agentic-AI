### Streaming

Receiving the Ai agent's response progressively in small chunks, instead of waiting for the complete answer to be generated. This improves the user experience for the longer responses. Use stream() with an iterator gives the output in chunks as they are produced.

```python
## Using Ollama (llama3.2) model :

from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")   ## we can use any other model of ollama also as per the requirements

## Without Streaming: Wait for the LLM to generate the enitre answer and then get the complete response
model.invoke("How is the whether today?")
```

```python
from langchain.agents import structured_output
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")

## With Streaming: Answer starts appering while LLM generating the response
for chunk in model.stream("How is the whether today?"):
    print(chunk.text, end = "|", flush = True)   ## end shows the chunks
```

```python
from langchain.agents import structured_output
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")

## With Streaming: 
for chunk in model.stream("Why are Elephants so big?"):
    print(chunk.text, end = "", flush = True)
```

### Batching

Batching means sending multiple inputs to an AI model together instead of processing each input separately. This improves performance and reduces costs as te processing can be done in parallel.

```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")

responses = model.batch([
    "What is the capital of India?",
    "Where is ney york?",
    "Who is the prime minister of India?"
])

for response in responses:
    print(response)
```

```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")

responses = model.batch([
    "What is the capital of India?",
    "Where is ney york?",
    "Who is the prime minister of India?"
    "Who is virat kohli?",
    "Tell me something about MS Dhoni."
    "How many continents are there?"
],
    config = {
        'max_concurrency': 5,      # Limits to 5 parallel calls
    }
)

for response in responses:
    print(response)
```
